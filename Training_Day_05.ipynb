**GIT & GITHUB**



**What is Git ?**
* Git is Software and Version Control System.
* Local Repository Server
* There are 3 Sections in Git
    * **Working Directory:** (We actually write and modify the code)
        * When we write the code and ready to upload it to github we fist need to add it to **Stretching Area** for that we use **git add .** command.
        * And to forward **Stretching Area** to github ready FIles we use **git commit -m "First Commit"** command and then to trigger this process we use **git push -u origin branchname** this sends file to github

**Why to use Git ?**
* Git Is used to store different data like our Project Data and our Application Data.
  and Maintain our History code.

**What is GitHub ?**
* It is a Central Repository Server
* Used to Store the data on the Browser or Web Server in and organized Manner

**Commands in Git**

* git init command    -->  Initialize
* git status command  -->  Check current status
* ls command          -->  Check's whether data is properly initialized or not
* add command         -->  Adds files (git add .)
* git commit command  -->  Save & Change (git commit -m "First Commit")
* push command        -->  Push data from git to github
* pull command        -->  Used to take the data from github to git  (git pull origin "branch-name")
* clone command       -->  Used to of the file
* pwd command         -->  To check the current directory where we are 

**How To Create Branch in Git ?**
* git branch -m branchname

**How to Create Branch in GitHub**
* Main Branch --> Right click --> New Branch

**How to Create New Repository ?**
* open github --> cick "New" --> Select "Repository" --> Give name to repository --> Optional description --> Allow "README.md" --> Click Create
