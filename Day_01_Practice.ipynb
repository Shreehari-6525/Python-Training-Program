**Q12. Find the sum and maximum**

Given the list below, find the sum and largest element without using the built-in sum() or max() functions.
* numbers = [12, 5, 28, 9, 16]
* Expected output: Sum = 70, Maximum = 28

```python
numbers = [12, 5, 28, 9, 16]
max = 0
sum = 0

for num in numbers:
    sum += num
    if num > max:
        max = num

print("Sum =", sum, "Maximum =", max)
```

**Q13. Separate even and odd numbers**

Given a list, create two separate lists: one containing even numbers and the other containing odd numbers.
* numbers = [1, 2, 3, 4, 5, 6, 7, 8]
* Expected output: Even = [2, 4, 6, 8], Odd = [1, 3, 5, 7]

```python
numbers = [1,2,3,4,5,6,7,8]

odd = []
even = []

for num in numbers:
    if num % 2 == 0:
        even.append(num)
    else:
        odd.append(num)

print("Even =", even, "Odd =", odd)
```

**Q14. Remove duplicate elements**

Convert the following list into a collection containing only unique elements.
* numbers = [2, 4, 2, 6, 4, 8, 8, 10]
* Expected unique values: 2, 4, 6, 8, 10

Challenge: Try solving it using a set, then try preserving the original order without using a set.

```python
numbers = [2, 4, 6, 4, 8, 8, 10]
duplicate_values = []
seen = set()

for num in numbers:
    if num in seen:
        duplicate_values.append(num)
    else:
        seen.add(num)


print(list(seen))
```

**Q15. Student record using a dictionary**
