**Single Inheritance**

```python
class product:

    def show(self):
        print("Name of Product:", self.name)
        print("Price of Product", self.price)

class IIT(product):
    def display(self):
        print("Welcome to IIT Mumbai")

obj = IIT()
obj.name = input("Enter Product Name")
obj.price = int(input("Enter Product Price"))
obj.display()
obj.show()
```

**Multi Level Inheritance**

* Contains 3 Classes
    1. Grand-Parent
    2. Parent
    3. Child

```python
class Department:
    def showDepartment(self):
        print("Your Department:", self.dept_name)

class Division(Department):
    def showDivision(self):
        print("Your Division:", self.division)

class Student(Division):
    def showStudent(self):
        print("Your Name:", self.name)

obj = Student()

obj.name = input("Enter Student Name")
obj.dept_name = "Computer Science"
obj.division = "B"

print("-------------- KSE DETAILS --------------")
obj.showStudent()
obj.showDivision()
obj.showDepartment()
```

**Multiple Inheritance**

* It contains 2 Parent Class and One Child Class
* Both Parent Class Inherit Single Child Class

```python
class parent1:
    def show(self):
        print("I am Mother")

class parent2:
    def display(self):
        print("I am Father")

class child(parent1, parent2):
    def showw(self):
        print("I am Child")


obj = child()

obj.show()
obj.display()
obj.showw()
```

**Hierarchal Inheritance**

* Contains 2 Child class and 1 Parent Class
* both child classes are inherited by single parent class

```python
class parent:
    def show(self):
        print("I am Parent")

class child1(parent):
    def displayChild1(self):
        print("I am Child1")

class child2(parent):
    def displayChild2(self):
        print("I am Child2")


obj1 = child1()
obj2 = child2()

obj1.show()
obj1.displayChild1()
print("---------------------------------")

obj2.show()
obj2.displayChild2()
```

**What is Polymorphism ?**

* A person with many roles, many forms, and many work

**Types of Polymorphism**

1. Compile Time Polymorphism (Method Overloading ( Same Function Name but Different Parameters ) is Used)
2. Run Time Polymorphism (Method Overriding ( Same Function name & Same Parameters but used in different classes ) is Used)



**Compile Time Polymorphism**

```python
class product:
    def show(self, name='', address=''):
        print('The Name of Product:', name)
        print("The Address of Product:", address)

obj = product()
obj.show()
obj.show('mobile', 'Pune')
```

```python
class product:
    def show(self):
        print("Name of Product", self.name)
        print("The Price of Product:", self.price)


class IIT(product):
    def show(self):
        super().show()
        print("Welcome to IIT Delhi")


obj = IIT()
obj.name = 'laptop'
obj.price = 50000
obj.show()
```

**What is Encapsulation ?**

* Wrapping of data and Method into single unit is called Encapsulation
* Eg We Mix Some Medicine in one Capsule
* Process of Binding States & Behavior of Object inside a single entity

**Access Modifier**

1. Public      self.name
2. Protected   self. _name
3. Private     self. _ _name

```python
class super:
    def __init__(self):
        self.__value1 = "This is Private 1"  # This is Private Variable
        self.__value2 = "This is Private 2"  # This is Private Variable

        self._value3 = "This is Protected 1"  # This is Protected Variable
        self._value4 = "Thsi is Protected 2"  # This is Protected Variable

    def show(self):
        print(self.__value1)
        print(self.__value2)

        print(self._value3)
        print(self._value4)


object = super()
object.show()
```

**What is Abstraction ?**

* Abstraction hide the internal data and Background detail and show only customer data

**Types of Abstractions**

1. Abstract Class
2. Interface

**Abstract v/s Interface**
* Abstract class store abstract method and also any other normal method then that method is available in body
* Interface store only abstract method and cannot store any normal methods this method is not available in the body

```python
from abc import ABC, abstractmethod

class mobile(ABC):
    @abstractmethod
    def printDetails(self):
        pass


class product(mobile):
    def printDetails(self):
        print("This is a Product")

class car(product):
    def printDetails(self):
        print("This is Car")


obj = car()
obj.printDetails()

obj = product()
obj.printDetails()
```

```python
from abc import ABC, abstractmethod

class product:
    @abstractmethod
    def printDetails(self):
        pass
    def mi(self):
        pass

class mobile(product):
    def printDetails(self):
        print("This is a Mobile")

class car(mobile):
    def m1(self):
        print("This is New Car")

obj = car()
obj.m1()
obj.printDetails()
```

**What is Stack ?**

* Stack is a Linear Data Structure
* Last In First Out Approach

```python
list = []

num = None
for i in range(1, 21):
    if i % 2 != 0:
        num = i * i
        list.append(num)
        num = None

print(list)
```

```python
dict = {}
num = None
for i in range(1, 21):
    if i % 2 != 0:
        num = i * i
        dict[i] = num

print(dict)
```
